### 7. Convert Celsius to Fahrenheit and Kelvin

```python
celsius = 37  # fixed temperature value in Celsius

# Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Convert Celsius to Kelvin
kelvin = celsius + 273.15

print(f"{celsius}°C = {fahrenheit}°F")
print(f"{celsius}°C = {kelvin}K")
```

### 15. Print grade based on student marks using if-elif-else

### 16. Nested if: check even AND greater than 100

```python
number = 104  # test value (change to test other cases)

# Outer if checks if the number is even
if number % 2 == 0:
    # Nested (inner) if checks if it is also greater than 100
    if number > 100:
        print(f"{number} is even and greater than 100")
    else:
        print(f"{number} is even but NOT greater than 100")
else:
    print(f"{number} is not even")
```

### 19. Sum of first 10 natural numbers using a while loop

```python
total = 0   # variable to hold the running sum
count = 1   # loop counter starting at 1

# Keep adding numbers until count exceeds 10
while count <= 10:
    total += count
    count += 1

print("Sum of first 10 natural numbers:", total)
```

### 20. Print even numbers between 1 and 50 using for loop and continue

```python
# Loop through numbers 1 to 50
for num in range(1, 51):
    if num % 2 != 0:
        continue  # skip odd numbers, move to the next iteration
    print(num)
```

### 21. Factorial of a number using a loop

```python
number = 6  # find factorial of this number
factorial = 1

# Multiply factorial by each number from 1 up to 'number'
for i in range(1, number + 1):
    factorial *= i

print(f"Factorial of {number} is {factorial}")
```

### 22. Print a right-angled star pattern using nested loops

```python
rows = 5  # number of rows in the pattern

# Outer loop controls the row number
for i in range(1, rows + 1):
    # Inner loop prints '*' i times on the same line
    for j in range(i):
        print("*", end="")
    print()  # move to the next line after each row
```

### 26. Union, intersection, and difference of two sets

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Union: all unique elements from both sets
print("Union:", set1 | set2)

# Intersection: elements common to both sets
print("Intersection:", set1 & set2)

# Difference: elements in set1 but not in set2
print("Difference (set1 - set2):", set1 - set2)
```
